```python
def count_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    lower_bound = Q1 - 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return len(outliers)
```

```python
print("Number of outliers per numerical column:")
for col in proj.select_dtypes(include=np.number).columns:
    num_outliers = count_outliers(proj, col)
    print(f"'{col}': {num_outliers}")
```

```python
def outlier_capping(proj, column):
 Q1 = proj[column].quantile(0.25)
 Q3 = proj[column].quantile(0.75)
 IQR = Q3 - Q1
 UE = Q3 + 1.5 * IQR
 LE = Q1 - 1.5 * IQR
 proj[column] = proj[column].apply(lambda x: UE if x > UE else LE if x < LE else x)

# Apply outlier capping to all numerical columns in the DataFrame 'proj'
for i in proj.select_dtypes(include=['int', 'float']).columns:
    outlier_capping(proj, i)
#treatment for all outlier columns
```

```python
target=proj[["Country"]]
features=proj.drop(columns=["Country"])
```

```python
from sklearn.preprocessing import StandardScaler
std=StandardScaler()
df=pd.DataFrame(std.fit_transform(features),columns=features.columns)
df.head()
```

```python
# As our dataset is large non hierarchial clustering is prefered
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

wcss = []
silhouette_scores = []
for i in range(2, 11):
    model = KMeans(n_clusters=i, init='k-means++' , random_state=42, n_init=10)
    model.fit(df)
    wcss.append(model.inertia_)

    silhouette_scores.append(silhouette_score(df,model.labels_))

# Plot the Elbow Method

plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
#By plotting we can see the Silhouette_score raising and dropping according k_means values
plt.figure(figsize=(6,3))
plt.plot(range(2,11),silhouette_scores,marker='o',
         color='red')
plt.title('Silhouette_score VS K-means')
plt.ylabel('silhouette_score')
plt.xlabel('k-means')
plt.show()

best_k_value = silhouette_scores.index(max(silhouette_scores)) + 2 # Corrected to add 2 for the actual k value
best_score1 = max(silhouette_scores)
print("best_k_value :",best_k_value)
print("best_score:",best_score1)
```

```python
# Choose an optimal number of clusters based on the elbow method
k = 3

kmeans = KMeans(n_clusters=k)
cluster_labels = kmeans.fit_predict(df)
print(kmeans.cluster_centers_)
print(kmeans.labels_)
print(kmeans.inertia_)
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
# Generate the linkage matrix
linked = linkage(df, method='ward') # 'ward' minimizes the variance within each cluster

# Plot the dendrogram
plt.figure(figsize=(15, 7))
dendrogram(linked,p=6,
           truncate_mode='level',
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distances')
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering

# Apply Agglomerative Clustering
hierarchical_model = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
hierarchical_labels = hierarchical_model.fit_predict(df)

# Add the hierarchical cluster labels to the DataFrame
df['Hierarchical_Cluster'] = hierarchical_labels

print('Hierarchical Cluster counts:')
print(df['Hierarchical_Cluster'].value_counts())
```

```python
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

# Re-run PCA to get the components for visualization if not already done with the full dataset
pca = PCA(n_components=2)
pca_values = pca.fit_transform(df)

# Visualize the hierarchical clusters using PCA
plt.figure(figsize=(10, 7))
sns.scatterplot(x=pca_values[:, 0], y=pca_values[:, 1], hue=df['Hierarchical_Cluster'], palette='Set1', s=60)
plt.title('Hierarchical Clustering Visualization (PCA)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

best_score3 = silhouette_score(df, hierarchical_labels)
print(f"The Silhouette Score for Agglomerative Clustering is: {best_score3}")
```

```python
from sklearn.mixture import GaussianMixture
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt

# Initialize and fit the Gaussian Mixture Model with 3 components
gmm = GaussianMixture(n_components=3, random_state=42, n_init=10)
gmm_labels = gmm.fit_predict(df)

# Add the GMM cluster labels to the DataFrame
df['GMM_Cluster'] = gmm_labels

print('GMM Cluster counts:')
print(df['GMM_Cluster'].value_counts())
```

```python
from sklearn.metrics import silhouette_score

best_score4 = silhouette_score(features, gmm_labels)
print(f"The Silhouette Score for Gaussian Mixture Model is: {best_score4}")
```

```python
%%writefile Model_Dep_app.py
import streamlit as st
import pickle
import pandas as pd
import numpy as np
import joblib
df=pd.read_csv('multi_cluster_data.csv')
@st.cache_resource
def load_models():
    kmeans = joblib.load("kmeans.pkl")
    gmm = joblib.load("gmm.pkl")
    hierarchical = joblib.load("hierarchial_model.pkl")
    birch = joblib.load("birch.pkl")
    return kmeans, gmm, hierarchical, birch
kmeans, gmm, hierarchical, birch = load_models()

# --------------------------------------------------
# Sidebar Selections
# --------------------------------------------------
method = st.sidebar.selectbox(
    "Select Clustering Method",
    ("KMEANS", "GMM", "BIRCH", "HIERARCHICAL")
)

country = st.sidebar.selectbox(
    "Select Country",
    df["Country"].unique()
)
status_map = {
    0: "🔴 Least Developed",
    1: "🟡 Developing",
    2: "🟢 Highly Developed"
}

country_data = df[df["Country"] == country]

X = country_data.drop(columns=[
    "Country",
    "GMM_Cluster",
    "Birch_Cluster",
    "Hierarchical_Cluster",
    "clust"
], errors="ignore")
if st.button("Predict Cluster"):

    if method == "KMEANS":
        cluster = kmeans.predict(X)[0]

    elif method == "GMM":
        cluster = gmm.predict(X)[0]

    elif method == "BIRCH":
        cluster = birch.predict(X)[0]

    elif method == "HIERARCHICAL":
        cluster = hierarchical.predict(X)[0]

    st.subheader(f"📍 {country} belongs to:")

    st.success(f"{status_map.get(cluster, 'Unknown')} (Cluster {cluster})")

```
